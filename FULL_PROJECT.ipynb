```python
df['gender'] = df['gender'].str.strip()
df['gender'] = df['gender'].str.lower()

df['department'] = df['department'].str.strip()
df['department'] = df['department'].str.lower()
```

```python
fig ,ax = plt.subplots(1,2,figsize=(16, 9))
df['gender'].value_counts().plot(kind='barh',ax=ax[0],xlabel='Value',ylabel='Gender',title='different values')
df['department'].value_counts().plot(kind='bar',ax=ax[1],xlabel='Department',ylabel='count',title='after unit form')
```

```python
df.loc[(df['age'] <= 0) | (df['age'] >= 100), 'age'] = np.nan
df.loc[(df['salary'] < 3000) , 'salary'] = np.nan
df.loc[(df['performance_score']> 10),'performance_score'] = np.nan
```

```python
x = df.isnull().sum().index
y = df.isnull().sum()
plt.figure(figsize=(18, 6))
plt.bar(x ,y )
plt.ylabel('Number Nulls')
plt.xlabel('Colums')
plt.title('Colums and Nulls')
plt.tight_layout()
```

```python

df =df.drop_duplicates()
```

```python

plt.pie([df.duplicated().sum(),len(df)],labels=['Dublicated data', 'all data'],colors=['red','blue'],autopct='%0.2f%%',shadow=True,explode=[0.9,0])
plt.title('Dublicate data')
```

```python
df.columns
```

```python
df['hire_date'] =pd.to_datetime(df['hire_date'],errors='coerce')
```

```python
df =df.dropna(subset=['gender'])
df =df.fillna({'department':'unknown','city':'unknown','email':'unknown'})
df['salary'] = df['salary'].fillna(
    df.groupby('department')['salary'].transform('median')
)
df['years_experience'] = df['years_experience'].fillna(
    df.groupby('department')['years_experience'].transform('median') )

df['performance_score'] = df['performance_score'].fillna(
    df.groupby('department')['performance_score'].transform('median'))
    
df['age'] =df['age'].fillna(df['age'].median()).round()
```

```python
x = df.isnull().sum().index
y = df.isnull().sum()
plt.figure(figsize=(18, 6))
plt.bar(x ,y )
plt.ylabel('Number Nulls')
plt.xlabel('Colums')
plt.title('Colums and Nulls')
plt.tight_layout()
```

```python
fig , ax = plt.subplots(2,3,figsize=(25,15))
ax[0,0].pie(df['department'].value_counts(),labels=df['department'].value_counts().index,autopct='%0.2f%%',shadow=True,explode=[0,0,0,0,0,0,0,0.6])
ax[0,0].set_title('Department Distribution')

df['age'].plot(kind='hist',ax=ax[0,1],bins =30)
ax[0,1].set_title('Age Distribution')
ax[0,1].set_xlabel('Age')

df['gender'].value_counts().plot(kind ='bar', ax =ax[0,2] )
ax[0,2].set_title('gender count')
ax[0,2].set_ylabel('count')

ax[1,0].scatter(df['years_experience'],df['performance_score'])
ax[1,0].set_title('is there connection between years_experience and performance_score  ')
ax[1,0].set_ylabel('performance_score')
ax[1,0].set_xlabel('years_experience')


y = df.groupby(df['department'])['salary'].mean()
x = df.groupby(df['department'])['salary'].mean().index
ax[1,1].plot(x ,y )
ax[1,1].set_title('mean salary by department')
ax[1,1].set_ylabel('salary')
ax[1,1].set_xlabel('Department')

df['Year'] = df['hire_date'].dt.year
df['Year'].value_counts().sort_index().plot(
    kind='line',
    ax=ax[1,2],
    marker='o'
)

ax[1,2].set_title('Employees Hired by Year')
ax[1,2].set_xlabel('Year')
ax[1,2].set_ylabel('Number of Employees')

fig.suptitle(
    'Employee Data Analysis Dashboard',
    fontsize=24,
    fontweight='bold'
)



```
